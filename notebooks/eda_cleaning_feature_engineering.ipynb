## 5–15. Cleaning, Feature Engineering, KPI Validation, Export

```python
df.columns = df.columns.str.strip().str.lower().str.replace(' ', '_')
for col in ['signup_date','last_active_date','churn_date']:
    df[col] = pd.to_datetime(df[col], errors='coerce')
df['churn_flag'] = df['churn'].astype(str).str.lower().isin(['1','true','yes','churned']).astype(int)
df['retained_flag'] = 1 - df['churn_flag']
df['engagement_band'] = pd.cut(df['engagement_score'], bins=[-1,35,70,101], labels=['Low Engagement','Medium Engagement','High Engagement'])
df['tenure_band'] = pd.cut(df['tenure_days'], bins=[-1,30,90,180,10000], labels=['0–30 Days','31–90 Days','91–180 Days','181+ Days'])
df['revenue_at_risk'] = df['revenue'] * df['churn_flag']
df['ltv_at_risk'] = df['lifetime_value'] * df['churn_flag']
df.to_csv('../data/processed/churn_retention_clean.csv', index=False)
df.head()
```

## Insight Summary
Low engagement, tenure bands, segment performance, and churn reasons are the main decision drivers for retention strategy.
